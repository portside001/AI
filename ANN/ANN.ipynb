```python
import pandas as pd
import numpy as np
import tensorflow as tf
```

```python
df = pd.read_csv('/Users/sachinrajput/Desktop/gen-ai-agentic-ai/AI/44_deep_learning_project/3rd, 4th -- ANN Theory , Installation/CHURN MODELING- TF_project/Churn_Modelling.csv')
```

```python
df.head()
```

```python
X = df.iloc[:,3:-1].values
# pd.DataFrame(X)
y = df.iloc[:,-1].values
```

```python
pd.DataFrame(X)
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
labelencoder = LabelEncoder()
X[:,2] = labelencoder.fit_transform(X[:,2])
```

```python
pd.DataFrame(X)
```

```python
from sklearn.compose import ColumnTransformer
ct = ColumnTransformer([('encoder', OneHotEncoder(), [1])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
```

```python
pd.DataFrame(X)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
pd.DataFrame(X)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.20, random_state=0)
```

```python
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

```python
# this is out first input layer 
ann = tf.keras.models.Sequential()
# We are creating a neural network model where layers will be added one after another in a sequence.
```

```python
# this is our hidden layer with 6 neurons and relu activation function
ann.add(tf.keras.layers.Dense(units=6, activation='relu'))
```

```python
# this is our second hidden layer with 5 neurons and relu activation function
ann.add(tf.keras.layers.Dense(units=5, activation='relu'))
```

```python
# this is our third hidden layer with 4 neurons and relu activation function
ann.add(tf.keras.layers.Dense(units=4, activation='relu'))
```

```python
# this is our fourth hidden layer with 3 neurons and relu activation function
ann.add(tf.keras.layers.Dense(units=3, activation='relu'))
```

```python
# output layer with 1 neuron and sigmoid activation function
ann.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))
```

till this step from input layer hidden layer and output layer we call as forward-prpogation

Forward propagation = making the prediction.
Backpropagation = learning from the error and adjusting weights.

```python
ann.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

```python
ann.fit(X_train,y_train,batch_size = 32,epochs = 150,validation_data=(X_test,y_test))
```

```python
y_pred = ann.predict(X_test)
```

```python
y_pred = (y_pred > 0.5)
```
